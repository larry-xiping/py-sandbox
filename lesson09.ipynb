⬜ Lesson 6 – Token Classification (NER)

NER stands for: Named Entity Recognition

````python
!pip install transformers -q
````

````python
from transformers import pipeline

# Initialize the NER (Named Entity Recognition) token classification pipeline
ner_pipeline = pipeline("token-classification")
````

````python
# Define your input text
text = "Jane works at Google and lives in Bartlesville, Oklahoma."

# Get the predictions
results = ner_pipeline(text)

# Print the results
for item in results:
    print(f"Word: {item['word']} | Label: {item['entity']} | Score: {item['score']:.4f}")
#Feed a string of text into the pipeline. The model will analyze each token and output its classification.
````

PER = Person
ORG = Organization
LOC = Location

I-PER
   ->
Inside a Person entity

I-ORG
   ->
Inside an Organization entity

I-LOC
   ->
Inside a Location entity

**Why Bartlesville is split**

That is because the model uses tokens, not always full words.

Bartlesville
->
Bart + ##les + ##ville

The ## means: this token is continuation of previous token

**NER only returns entities**

The purpose of NER is: Sentence
 ->
Find important named entities. works/at/and are just normal words. So the model ignores them.

````python
#You are not restricted to the default model. You can specify highly accurate or domain-specific models directly.

# Example: Use a large biomedical NER model
bio_pipeline = pipeline("token-classification", model="d4data/biomedical-ner-all")

bio_text = "The patient was treated with Nivolumab for advanced melanoma."
print(bio_pipeline(bio_text))

# Get the predictions
results =  bio_pipeline(bio_text)

# Print the results
for item in results:
    print(f"Word: {item['word']} | Label: {item['entity']} | Score: {item['score']:.4f}")
````

````python
from transformers import pipeline
import numpy as np

extractor = pipeline("feature-extraction")

v1 = extractor("I love Python")
v2 = extractor("I enjoy programming")
````

v1
->
List of sentences

v1[0]
->
List of tokens

v1[0][0]
->
List of embedding values

v1[0][0][0]
->
Float

```
v1 =
[
    [                           # Sentence
        [0.12, -0.45, ...],     # Token 1 embedding
        [0.87,  0.34, ...],     # Token 2 embedding
        [-0.21, 0.56, ...]      # Token 3 embedding
    ]
]
```



````python
v1 = np.array(v1[0]).mean(axis=0)
v2 = np.array(v2[0]).mean(axis=0)

distance = np.linalg.norm(v1 - v2)

print(distance)
````

````python
v3 = extractor("Banana smoothie recipe")
v3 = np.array(v3[0]).mean(axis=0)

print(np.linalg.norm(v1 - v3))
````

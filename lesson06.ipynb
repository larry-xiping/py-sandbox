Pandas -> Matplotlib -> Chart

When you call df.plot(), Pandas acts as a convenient front-end for Matplotlib. Pandas takes the data stored in the DataFrame, automatically converts it into chart instructions, and then passes those instructions to Matplotlib, which actually draws and displays the chart. In other words, Pandas makes plotting easy by providing a simple plot() method, while Matplotlib is the underlying graphics engine that creates the visual output you see on the screen.

➡️ Lesson 3 – Multiple Lines

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

revenue = [100, 120, 90, 150]

cost = [70, 80, 75, 100]
```

```python
plt.plot(months, revenue)     #Draw Two Lines

plt.plot(months, cost)

plt.show()
```

```python
plt.plot(
    months,
    revenue,
    label="Revenue"     #Add Labels
)

plt.plot(
    months,
    cost,
    label="Cost"
)

plt.show()    #The labels exist internally but are not displayed yet.
```

```python
plt.plot(
    months,
    revenue,
    label="Revenue"     #defines the name
)

plt.plot(
    months,
    cost,
    label="Cost"
)

plt.legend()      #Add Legend, Now a legend appears. displays the names
plt.show()
```

```python
plt.plot(
    months,
    revenue,
    marker="o",     #Add Markers
    label="Revenue"
)

plt.plot(
    months,
    cost,
    marker="s",
    label="Cost"
)

plt.legend()

plt.show()
```

**Common markers:**

o  circle

s  square

^  triangle

*  star

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

revenue = [100, 120, 90, 150]

cost = [70, 80, 75, 100]

plt.figure(figsize=(8,4))

plt.plot(
    months,
    revenue,
    marker="o",
    label="Revenue"
)

plt.plot(
    months,
    cost,
    marker="s",
    label="Cost"
)

plt.title("Revenue vs Cost")

plt.xlabel("Month")

plt.ylabel("Amount")

plt.grid(True)

plt.legend()

plt.show()
```

➡️ Lesson 4 – Bar Charts

```python
import matplotlib.pyplot as plt

systems = ["ECC", "S4", "MDG", "BW"]

tickets = [12, 8, 5, 3]
```

```python
plt.bar(
    systems,
    tickets
)

plt.show()
```

```python
plt.bar(
    systems,
    tickets
)

plt.title("Tickets by System")      #Add Title and Labels

plt.xlabel("System")

plt.ylabel("Ticket Count")

plt.show()
```

```python
plt.bar(
    systems,
    tickets,
    color="skyblue"     #Add Color
)

plt.show()
```

```python
plt.bar(
    systems,
    tickets,
    color=[     #Different Colors
        "blue",
        "green",
        "orange",
        "red"
    ]
)

plt.show()
```

```python
plt.bar(
    systems,
    tickets
)

plt.title("Tickets by System")

plt.ylabel("Ticket Count")

plt.grid(True)      #Add Grid Lines

plt.show()
```

```python
plt.barh(     #Horizontal Bar Chart
    systems,
    tickets
)

plt.title("Tickets by System")

plt.show()
```

➡️ Lesson 5 – Pie Charts

```python
import matplotlib.pyplot as plt

departments = [
    "IT",
    "HR",
    "Finance",
    "Operations"
]

employees = [
    25,
    10,
    15,
    20
]
```

```python
plt.pie(employees)

plt.show()
```

```python
plt.pie(
    employees,
    labels=departments
)

plt.show()
```

```python
plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%"   #Show Percentages
)

plt.show()
```

```python
plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%"
)

plt.title("Employee Distribution")    #Add a Title

plt.show()
```

```python
plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%",
    colors=[
        "blue",
        "green",
        "orange",
        "red"
    ]
)

plt.title("Employee Distribution")

plt.show()
```

```python
plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%",
    explode=[0.1,0,0,0]     #Explode a Slice
)

plt.show()
```

```python
import matplotlib.pyplot as plt

scores = [65, 70, 72, 75, 78, 80, 81, 85, 88, 90, 92, 95]

plt.hist(scores)

plt.title("Exam Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.hist(scores, bins=10)     #More bins:
```

```python
plt.hist(scores,
         bins=5,
         color="green")

plt.show()
```

```python
plt.hist(scores,
         bins=5,
         color="skyblue",
         edgecolor="black")     #Add Border Lines

plt.show()
```

✅ Lesson 7 – Scatter Plots

Scatter Plots are one of the most important charts in data analysis because they help you see relationships between two numeric variables.

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
scores = [60, 70, 75, 85, 95]

plt.scatter(hours, scores)

plt.title("Study Hours vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.show()
```

```python
plt.scatter(hours,
            scores,
            color="red")      #Change Color

plt.show()
```

```python
plt.scatter(hours,
            scores,
            s=200)      #Change Marker Size

plt.show()
```

```python
plt.scatter(hours,
            scores,
            marker="*")     # common makers "o"   circle, "s"   square, "^"   triangle, "*"   star

plt.show()
```

➡️ Lesson 8 – Subplots

Subplots allow you to place multiple charts inside one figure.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [10, 20, 30, 40]
y2 = [40, 30, 20, 10]

plt.subplot(1, 2, 1)      #1 row, 2 columns, Position 1
plt.plot(x, y1)
plt.title("Chart 1")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("Chart 2")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.subplot(2, 2, 1)
plt.plot([1,2,3])

plt.subplot(2, 2, 2)
plt.bar(["A","B","C"], [5,8,6])

plt.subplot(2, 2, 3)
plt.pie([40,30,30])

plt.subplot(2, 2, 4)
plt.hist([1,2,2,3,3,3,4,5])

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Sales": [100,120,140,160],
    "Profit": [20,25,30,35]
})

fig, ax = plt.subplots(1, 2)

ax[0].plot(df["Sales"])
ax[0].set_title("Sales Trend")

ax[1].plot(df["Profit"])
ax[1].set_title("Profit Trend")

plt.show()
```

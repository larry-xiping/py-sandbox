**NumPy (Numerical Python)** is a Python library that provides a high-performance, multidimensional array data structure and mathematical operations on those arrays. The core object is *numpy.ndarray* where ndarray = N-dimensional Array.

- Python list stores each element as a separate Python object
  - flexible but slower and uses more memory
- NumPy stores numbers in continuous memory  
  - Faster; More memory efficient; Better for math

```python
import numpy as np

ages = np.array([30, 25, 40])

type(ages)    #ndarray: N-Dimensional Array
```

```python
ages1 = [30, 25, 40]
type(ages1)
```

```python
print(ages[0])
type(ages[0])
```

```python
print(ages1[0])
type(ages1[0])
```

```python
ages = np.array([30, 25, 40])
ages + 10
```

```python
[30, 25, 40] + [10]  #List
```

```python
ages.mean()
```

➡️ Lesson 4 – Array Filtering

```python
import numpy as np

ages = np.array([30, 25, 40, 35, 20])

ages
```

```python
ages > 30 #Create a Condition
```

```python
ages[ages > 30] #Use the Condition, called Boolean Filtering, keeps only the True values.
```

```python
ages[ages < 30]
```

```python
ages[ages == 30]
```

```python
ages[ages != 30]
```

```python
ages[(ages > 25) & (ages < 40)]   #Multiple Conditions (AND); alway use (condition1) & (condition2)
```

```python
ages[(ages < 25) | (ages > 35)] #Multiple Conditions (OR)
```

```python
data = np.array([
    [30, 100000],
    [25,  90000],
    [40, 120000]
])

data
```

```python
data[:,0] > 30
```

```python
data[data[:,0] > 30]
```

➡️ Lesson 5 – Mathematical Operations

```python
import numpy as np

salary = np.array([
    100000,
     90000,
    120000
])
```

```python
salary * 1.10     #Vectorized Operations, NumPy : processes the whole array at once
```

```python
bonus = np.array([
    5000,
    4000,
    6000
])

salary + bonus    #Array + Array
```

```python
salary.mean()   #Statistics
```

```python
salary.max()
```

```python
salary.std()
```

```python
data = np.array([
    [30,100000],
    [25, 90000],
    [40,120000]
])
```

```python
data + 10   #Matrix Math
```

```python
data[:,1]
```

```python
data[:,1].mean()
```

➡️ Lesson 6 – Random Numbers

```python
import numpy as np

np.random.rand()    #Generate One Random Number, 0 ≤ number < 1
```

```python
np.random.rand(5)   #Generate Multiple Random Numbers
```

```python
np.random.randint(18, 66, 5)    #Generate Random Integers
```

```python
np.random.randint(    #Generate a Matrix
    1,
    100,
    (3,2)
)
```

```python
np.random.seed(42)    #Set a Random Seed, To make results repeatable
```

```python
np.random.randint(1,100,5)    #always produces the same numbers.
```

```python
employees = np.array([
    "David",
    "Alice",
    "Bob",
    "Carol"
])
```

```python
np.random.choice(employees)   #Random Choice, Pick one person
```

```python
np.random.choice(     #Pick two people
    employees,
    2
)
```

```python
rolls = np.random.randint(      #Simulation Example,Roll a die 10 times
    1,
    7,
    10
)

rolls
```

```python
rolls.mean()
```

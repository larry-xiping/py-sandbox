```python
class Employee:

    def __init__(self, name, age):
        self.name = name
        self.age = age

emp1 = Employee("David", 30)

print(emp1.name)
print(emp1.age)
```

###dunder methods
- __init__()  → called when creating an object
- __str__()   → called when printing an object
- __len__()   → called by len()


```python
class Employee:

    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello", self.name)

emp = Employee("David")

emp.greet()
```

```python
class Ticket:

    def __init__(self, number, status):
        self.number = number
        self.status = status

    def display(self):
        print(self.number, self.status)

t1 = Ticket("INC1001", "Open")
t2 = Ticket("INC1002", "Closed")

t1.display()
t2.display()
```

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print("My name is", self.name)
        print("I am", self.age, "years old")

david = Person("David", 30)

david.introduce()
```

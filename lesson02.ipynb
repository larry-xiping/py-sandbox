## Phase 2: Core Containers
- Variable  = one value
- List = many values
- Dictionary  = key/value pairs

```python
systems = ["ECC", "S4", "MDG"]

print(systems)
```

```python
systems = ["ECC", "S4", "MDG"]

print(systems[0])
print(systems[1])
print(systems[2])
```

```python
systems = ["ECC", "S4"]

systems.append("MDG")

print(systems)
print(len(systems))
```

```python
systems = ["ECC", "S4", "MDG"]

for system in systems:
    print(system)

print(type(system))
print(type(systems))
```

```python
employee = {
    "name": "David",
    "title": "Development Integrator",
    "age": 30
}

print(employee)
print(employee["name"])
```

```python
employee["company"] = "Google"

print(employee)
```

```python
for key, value in employee.items():
  print(key, ":", value)
```

```python
vendors = [
    {"vendor": "100001", "name": "ABC Oil"},
    {"vendor": "100002", "name": "XYZ Energy"},
    {"vendor": "100003", "name": "Global Supply"}
]
for vendor in vendors:
    print(vendor["vendor"], vendor["name"])

print(type(vendors))
print(type(vendor["vendor"]))
```

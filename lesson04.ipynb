⬜ Lesson 3 – Filtering

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["David", "Alice", "Bob"],
    "Age": [30, 25, 40],
    "Salary": [100000, 90000, 120000]
})

df
```

```python
df[df["Age"] > 30]    #Filter Age > 30
```

```python
df[df["Salary"] >= 100000]  #Filter Salary >= 100000
```

```python
df[df["Name"] == "David"] #Filter by Name
```

```python
df[df["Name"] != "David"] #Not Equal
```

```python
df[                             #Multiple Conditions  (condition1) & (condition2)
    (df["Age"] > 25)
    &
    (df["Salary"] > 100000)
]
```

```python
df[                           #OR Condition   (condition1) | (condition2)
    (df["Age"] < 30)
    |
    (df["Salary"] > 100000)
]
```

➡️ Lesson 5 – Grouping & Aggregation

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["David", "Alice", "Bob", "Carol"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [100000, 90000, 120000, 95000]
})

df
```

```python
df.groupby("Department")["Salary"].mean()   #Average Salary by Department
                                            #SELECT Department,AVG(Salary)
                                            #FROM Employee
                                            #GROUP BY Department
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
df.groupby("Department")["Name"].count()
```

```python
df.groupby("Department")["Salary"].max()
```

```python
df.groupby("Department")["Salary"].agg(       #Multiple Aggregations
    ["count", "sum", "mean", "max", "min"]
)
```

```python
import pandas as pd

df = pd.DataFrame({                                 #group by multiple fields
    "Department": ["IT", "IT", "HR", "HR", "IT"],
    "Location": ["Tulsa", "Houston", "Tulsa", "Houston", "Tulsa"],
    "Salary": [90000, 95000, 70000, 75000, 100000]
})

df.groupby(["Department", "Location"])["Salary"].mean()
```

```python
df.groupby(["Department", "Location"])["Salary"].mean().reset_index()  #reset_index() is used to turn the grouping columns back into regular columns
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["David", "Alice", "Bob"],
    "Age": [30, None, 40],
    "Salary": [100000, 90000, None]
})

df  #NaN Not a Number
```

```python
df["Salary"].fillna(0)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["David", "Alice", "Bob"],
    "Age": [30, None, 40],
    "Salary": [100000, 90000, None]
})

df  #NaN Not a Number
```

```python
df["Age"].fillna(df["Age"].mean())  #Fill Using Average. This is called imputation.
```

```python
df  #fillna() returns a new result. It does not automatically modify the DataFrame.
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())  #Now Age is permanently updated.
df
```

⬜ Lesson 9 – Joining Tables

```python
import pandas as pd

employees = pd.DataFrame({    #Create Table 1
    "EmployeeID": [1, 2, 3],
    "Name": ["David", "Alice", "Bob"]
})

employees
```

```python
salaries = pd.DataFrame({   #Create Table 2
    "EmployeeID": [1, 2, 3],
    "Salary": [100000, 90000, 120000]
})

salaries
```

```python
result = pd.merge(    #Join the Tables
    employees,
    salaries,
    on="EmployeeID"
)

result
```

```python
employees = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Name": ["David", "Alice", "Bob"]
})

salaries = pd.DataFrame({
    "EmployeeID": [1, 2],
    "Salary": [100000, 90000]
})

pd.merge(   #LEFT JOIN
    employees,
    salaries,
    on="EmployeeID",
    how="left"
)
```

```python
employees = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Name": ["David", "Alice", "Bob"]
})

salaries = pd.DataFrame({
    "EmployeeID": [1, 2],
    "Salary": [100000, 90000]
})

pd.merge(   #RIGHT JOIN
    employees,
    salaries,
    on="EmployeeID",
    how="right"
)
```

```python
import pandas as pd

employees = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Name": ["David", "Alice", "Bob"]
})

salaries = pd.DataFrame({
    "EmployeeID": [2, 3, 4],
    "Salary": [90000, 120000, 150000]
})

result = pd.merge(    #Outer Join
    employees,
    salaries,
    on="EmployeeID",
    how="outer"
)

result
```

✅ Lesson 10 – Data Visualization

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "Operations"],
    "Employees": [25, 10, 15, 20],
    "Budget": [500000, 200000, 350000, 400000]
})

df
```

```python
df.plot(
    x="Department",
    y="Employees",
    kind="bar"    #Create a Bar Chart
)
```

```python
df.plot(
    x="Department",
    y="Employees",
    kind="bar",
    title="Employees by Department"
)
```

```python
sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Revenue": [100000, 120000, 90000, 150000]
})

sales
```

```python
sales.plot(
    x="Month",
    y="Revenue",
    kind="line",    #Line Chart
    title="Monthly Revenue"
)
```

```python
df.plot(
    y="Employees",
    kind="pie",
    labels=df["Department"],
    autopct="%1.1f%%",    #"%1.0f%%" ->40%; "%1.2f%%" ->40.00%
    title="Employee Distribution"
)
```

```python
df.plot(
    x="Department",
    y="Budget",
    kind="barh",    #Horizontal Bar Chart
    title="Budget by Department"
)
```

##Common Chart Types
- bar     = compare categories
- barh    = compare categories horizontally
- line    = show trend
- pie     = show percentage/share
- hist    = show distribution
- scatter = show relationship

```python
employees = pd.DataFrame({
    "Name": ["David", "Alice", "Bob", "Carol"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [100000, 90000, 120000, 95000]
})

avg_salary = employees.groupby("Department")["Salary"].mean()

avg_salary.plot(
    kind="bar",
    title="Average Salary by Department"
)
```

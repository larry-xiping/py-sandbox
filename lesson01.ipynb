Python does not use braces. Instead, it uses indentation (spaces at the beginning of a line) to define blocks. After lines ending with a colon (:), Python expects an indented block. Most editors (including Colab) automatically insert the 4 spaces when you press Tab. You can think of indentation in Python as serving the same purpose as {} block delimiters in C, C++, Java, and JavaScript.

```python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("C")
```

Python is dynamically typed, so you do not declare variables before using them. The same variable name can refer to objects of different types over time.

```python
for i in range(5):
    print(i)
```

Python generally does not require a special end-of-line symbol. Each newline marks the end of a statement. If Python sees an open parenthesis, bracket, or brace, it knows the statement isn't finished yet.

range(start, stop, step). The stop value is NOT included.

```python
x = 1

while x <= 5:
    print(x)
    x += 1
```

⬜ JSON Files

Think of JSON as the Excel file format of the internet.

```python
employee = {
    "Name": "David",
    "Age": 30,
    "Salary": 100000
}

employee      #Create a Python Dictionary
```

```python
import json

employee = {
    "Name": "David",
    "Age": 30,
    "Salary": 100000
}
#Save as JSON
  #Open the file.
  #Assign the file object to variable f.
  #Execute the code inside the block.
  #Automatically close the file when the block ends.
with open("employee.json", "w") as f:
    json.dump(employee, f)
```

```python
import json

with open("employee.json", "r") as f:
    data = json.load(f)     #Read JSON File

data
```

```python
data["Name"]    #Access Fields
```

```python
import pandas as pd

df = pd.DataFrame([data])     #Convert JSON to DataFrame

df
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["David", "Alice", "Bob", "John"],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Salary": [100000, 90000, 120000, 110000]
})
#Create Sample Employee Data
df.to_csv("employees.csv", index=False)
```

```python
df = pd.read_csv("employees.csv")

df
```

```python
df["Salary"].mean()     #Calculate Average Salary
```

```python
high_salary = df[df["Salary"] > 100000]     #Find High-Paid Employees

high_salary
```

```python
dept_summary = df.groupby("Department")["Salary"].mean()    #Group by Department

dept_summary
```

```python
dept_summary.to_excel("salary_summary.xlsx")    #Save Results to Excel
```

```python
import matplotlib.pyplot as plt

dept_summary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.show()      #Create a Chart
```
